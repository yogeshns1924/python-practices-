```python
y=[]

for i in range(10):
  num=int(input("enter the num:"))
  y.append(num)

print("y:",y)

print("even number:")
for nums in y:
  if nums%2==0:
    print(nums)

print("odd number:")
for nums in y:
  if nums%2!=0:
    print(nums)

maximum=max(y)
minmum=min(y)

print("max:",maximum)
print("min:",minmum)
```
