```python
books = ["Python", "SQL", "Java"]

choice = input("Do you want a book? (yes/no): ")

if choice == "yes":
    card = input("Do you have a library card? (yes/no): ")

    if card == "yes":
        print("Books:", books)

        book = input("Enter the book name: ")

        if book in books:
            books.remove(book)
            print("Book issued successfully.")
            print("Remaining books:", books)
        else:
            print("Book not available.")
    else:
        print("Library card is required.")
else:
    print("Thank you!")
```

```python
num = int(input("Enter a number: "))

temp = num
sum = 0

while temp > 0:
    digit = temp % 10
    sum = sum + digit ** 3
    temp = temp // 10

if sum == num:
    print("Armstrong Number")
else:
    print("Not an Armstrong Number")
```

```python
def unique_list(numbers):
    unique = []

    for i in numbers:
        if i not in unique:
            unique.append(i)

    return unique


sample = [1, 2, 3, 3, 3, 3, 4, 5]
print(unique_list(sample))
```

```python
def count_vowels(text):
    count = 0

    for i in text.lower():
        if i in "aeiou":
            count += 1

    return count


word = input("Enter a string: ")
print("Number of vowels:", count_vowels(word))
```

```python
def is_prime(n):
    if n <= 1:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True


num = int(input("Enter a number: "))

if is_prime(num):
    print("Prime Number")
else:
    print("Not a Prime Number")
```

```python
def calculator(a, b, op):
    if op == "+":
        return a + b
    elif op == "-":
        return a - b
    elif op == "*":
        return a * b
    elif op == "/":
        return a / b
    else:
        return "Invalid Operator"


num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
operator = input("Enter operator (+, -, *, /): ")

print("Result =", calculator(num1, num2, operator))
```
